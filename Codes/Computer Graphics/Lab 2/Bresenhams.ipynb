```python
x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))
x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))

dx = x2 - x1
dy = y2 - y1

p0 = 2 * dy - dx
```

```python
xi = x1
yi = y1

print(f"Point: ({xi}, {yi})")
while True:
     if p0 < 0:
          p0 += 2 * dy
     else:
          p0 += 2 * dy - 2 * dx
          yi += 1
     xi += 1
     print(f"Point: ({xi}, {yi})")
     if xi == x2 and yi == y2:
          break
```

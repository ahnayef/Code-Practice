```python
def dda(x1, y1, x2, y2):
     dx = x2 - x1
     dy = y2 - y1

     steps = max(abs(dx), abs(dy))

     x_inc = dx / steps
     y_inc = dy / steps
     print("Pixels:")
     for i in range(steps + 1):
          print(f"({round(x1)}, {round(y1)})")
          # dda_px+=1
          x1 += x_inc
          y1 += y_inc
```

```python
def bresenham(x1, y1, x2, y2):
     tmp_x1=x1
     tmp_y1=y1
     tmp_x2=x2
     tmp_y2=y2

     flag = (x1>x2)
     # print(flag)
     if(flag):
          x1=tmp_y1
          y1=tmp_x1
          x2=tmp_y2
          y2=tmp_x2


     dx = x2 - x1
     dy = y2 - y1

     p0 = (2 * dy) - dx
     xi = x1
     yi = y1
     print("Pixels:")
     bresenham_px+=1
     if(flag):
          print(f"({yi}, {xi})")
     else:
          print(f"({xi}, {yi})")
     
     while ((xi != x2) and (yi != y2)):
          if p0 < 0:
               p0 += (2 * dy)
          else:
               p0 += (2 * dy) - (2 * dx)
               yi += 1
          xi += 1
          if(flag):
               print(f"({yi}, {xi})")
          else:
               print(f"({xi}, {yi})")
          print((xi != x2) and (yi != y2))
          bresenham_px+=1


```

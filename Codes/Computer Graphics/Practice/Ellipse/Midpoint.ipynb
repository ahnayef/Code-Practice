```python
def midpoint_ellipse(xc, yc, rx, ry):
    x = 0
    y = ry

    p1 = (ry ** 2) - (rx ** 2 * ry) + (0.25 * rx ** 2)
    
    dx = 2 * ry ** 2 * x
    dy = 2 * rx ** 2 * y

    # Region 1
    while dx < dy:
        add_symmetric_points(xc, yc, x, y)
        
        x += 1
        dx += 2 * ry ** 2
        
        if p1 < 0:
            p1 += ry ** 2 + dx
        else:
            y -= 1
            dy -= 2 * rx ** 2
            p1 += ry ** 2 + dx - dy
    # Region 2
    p2 = ((ry ** 2) * ((x + 0.5) ** 2)) + ((rx ** 2) * ((y - 1) ** 2)) - (rx ** 2 * ry ** 2)

    while y >= 0:
        add_symmetric_points(xc, yc, x, y)
        
        y -= 1
        dy -= 2 * rx ** 2
        
        if p2 > 0:
            p2 += rx ** 2 - dy
        else:
            x += 1
            dx += 2 * ry ** 2
            p2 += rx ** 2 - dy + dx

    return points

```

```python
def dda_line(a, b, c, d):
     dx = c - a
     dy = d - b
     steps = max(abs(dx), abs(dy))
     print(steps)
     x_inc = dx / steps
     y_inc = dy / steps
     x, y = a, b
     points = []
     while steps > 0:
          points.append((round(x), round(y)))
          x += x_inc
          y += y_inc
          steps -= 1
     return points

x1= int(input("x1:"))
y1= int(input("y1:"))
x2= int(input("x2:"))
y2= int(input("y2:"))
x3= int(input("x3:"))
y3= int(input("y3:"))
x4= int(input("x4:"))
y4= int(input("y4:"))

p1 = (x1, y1)
p2 = (x2, y2)
p3 = (x3, y3)
p4 = (x4, y4)

print(dda_line(p1[0],p1[1], p2[0], p2[1]))
```

```python
def bresenham(x1, y1, x2, y2):
    points = []
    dx = abs(x2 - x1)
    dy = abs(y2 - y1)
    p0= 2*dy - dx
    x, y = x1, y1
    while x != x2 or y != y2:
        points.append((x, y))
        if p0 < 0:
            p0 += 2 * dy
            x += 1 if x1 < x2 else -1
        else:
            p0 += 2 * dy - 2*dx
            y += 1 if y1 < y2 else -1

    return points

print(bresenham(p1[0],p1[1], p2[0], p2[1]))
```
